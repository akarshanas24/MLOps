```python
import json
import os

affected_folder = "/content/drive/MyDrive/DATASET/affected_frames"

label_output = "/content/yolo_labels"

os.makedirs(label_output, exist_ok=True)

for file in os.listdir(affected_folder):

    if not file.endswith(".json"):
        continue

    with open(
        os.path.join(affected_folder, file),
        "r"
    ) as f:

        data = json.load(f)

    img_w = data["imageWidth"]
    img_h = data["imageHeight"]

    txt_name = file.replace(".json", ".txt")

    txt_path = os.path.join(
        label_output,
        txt_name
    )

    with open(txt_path, "w") as out:

        for shape in data["shapes"]:

            points = shape["points"]

            xs = [p[0] for p in points]
            ys = [p[1] for p in points]

            xmin = min(xs)
            xmax = max(xs)

            ymin = min(ys)
            ymax = max(ys)

            x_center = ((xmin+xmax)/2)/img_w
            y_center = ((ymin+ymax)/2)/img_h

            width = (xmax-xmin)/img_w
            height = (ymax-ymin)/img_h

            out.write(
                f"0 {x_center} {y_center} {width} {height}\n"
            )

print("YOLO Labels Created")
```

```python
from sklearn.model_selection import train_test_split
import os
import shutil

affected_folder = "/content/drive/MyDrive/DATASET/affected_frames"
normal_folder = "/content/drive/MyDrive/DATASET/normal_frames"

affected_labels = "/content/yolo_labels"
normal_labels = "/content/normal_labels"

dataset_root = "/content/drive/MyDrive/Healthcare_Detection_Dataset"

# ----------------------------------
# Collect affected images
# ----------------------------------

affected = []

for file in os.listdir(affected_folder):

    if file.endswith(".jpg"):

        affected.append(
            (
                os.path.join(affected_folder,file),
                os.path.join(
                    affected_labels,
                    file.replace(".jpg",".txt")
                )
            )
        )

# ----------------------------------
# Collect normal images
# ----------------------------------

normal = []

for file in os.listdir(normal_folder):

    if file.endswith(".jpg"):

        normal.append(
            (
                os.path.join(normal_folder,file),
                os.path.join(
                    normal_labels,
                    file.replace(".jpg",".txt")
                )
            )
        )

print("Affected:",len(affected))
print("Normal:",len(normal))
```

```python
import os

affected_folder = "/content/drive/MyDrive/DATASET/affected_frames"
label_folder = "/content/yolo_labels"

valid_images = []

for img in os.listdir(affected_folder):

    if not img.endswith(".jpg"):
        continue

    label_name = img.replace(".jpg", ".txt")

    if os.path.exists(
        os.path.join(label_folder, label_name)
    ):
        valid_images.append(img)

print("Valid Affected Images =", len(valid_images))
```

```python
import os
import random

normal_folder = "/content/drive/MyDrive/DATASET/normal_frames"

normal_images = [
    x for x in os.listdir(normal_folder)
    if x.endswith(".jpg")
]

random.seed(42)

normal_images = random.sample(
    normal_images,
    332
)

print(len(normal_images))
```

```python
from sklearn.model_selection import train_test_split

train_aff, temp_aff = train_test_split(
    valid_images,
    test_size=0.30,
    random_state=42
)

val_aff, test_aff = train_test_split(
    temp_aff,
    test_size=0.50,
    random_state=42
)

train_norm, temp_norm = train_test_split(
    normal_images,
    test_size=0.30,
    random_state=42
)

val_norm, test_norm = train_test_split(
    temp_norm,
    test_size=0.50,
    random_state=42
)

print(len(train_aff), len(val_aff), len(test_aff))
print(len(train_norm), len(val_norm), len(test_norm))
```
