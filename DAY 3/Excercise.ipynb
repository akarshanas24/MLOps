LINEAR REGRESSION

```python
# ==========================================
# LINEAR REGRESSION - CANADA PER CAPITA INCOME
# ==========================================

# Import Required Libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Read CSV File
df = pd.read_csv("/content/canada_per_capita_income.csv")

# Display Dataset
print(df.head())

# Independent Variable
X = df[['year']]

# Dependent Variable
y = df['per capita income (US$)']

# Create Model
model = LinearRegression()

# Train Model
model.fit(X, y)

prediction = model.predict([[2020]])

# Print Prediction
print("\nPredicted Per Capita Income in 2020:")
print(prediction[0])

# Slope
print("\nCoefficient:")
print(model.coef_)

# Intercept
print("\nIntercept:")
print(model.intercept_)

# Scatter Plot
plt.scatter(df['year'], df['per capita income (US$)'])

# Regression Line
plt.plot(df['year'], model.predict(X))

# Labels
plt.xlabel("Year")
plt.ylabel("Per Capita Income (US$)")
plt.title("Linear Regression - Canada Income Prediction")

# Show Plot
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
from word2number import w2n
```

```python
d = pd.read_csv("hiring.csv")
d
```

```python
d.experience = d.experience.fillna("zero")
d
```

```python
d.experience = d.experience.apply(w2n.word_to_num)
d
```

```python
import math
median_test_score = math.floor(d['test_score(out of 10)'].mean())
median_test_score
```

```python
d['test_score(out of 10)'] = d['test_score(out of 10)'].fillna(median_test_score)
d
```

```python
reg = linear_model.LinearRegression()
reg.fit(d[['experience','test_score(out of 10)','interview_score(out of 10)']],d['salary($)'])
```
